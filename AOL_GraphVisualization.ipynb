```r
library(tidyverse)
```

```r
data <- read.csv("diabetes_data.csv")
head(data)
```

```r
hist(data$Age, main = "Distribusi Umur", xlab = "Umur", ylab = "Jumlah", col = "#98F5F9")
```

```r
ggplot(data, aes(x = class, fill = class)) +
  geom_bar() + xlab("Class") + ylab("Jumlah") + ggtitle("Rasio Resiko Diabetes")
```

```r
ggplot(data, aes(x = class, y = Age)) +
  geom_boxplot(fill = "blue", color = "black") +
  labs(title = "Class and Age Relationship", x = "Class", y = "Age") +
  theme_minimal()
```
